```python
import pandas as pd
```

```python
ab_depth2 = pd.read_csv("../Evaluation_Data/AlphaBeta/ab_d2.csv")
ab_depth4 = pd.read_csv("../Evaluation_Data/AlphaBeta/ab_d4.csv")
ab_depth6 = pd.read_csv("../Evaluation_Data/AlphaBeta/ab_d6.csv")
mm_depth2 = pd.read_csv("../Evaluation_Data/Minimax/mm_d2.csv")
mm_depth4 = pd.read_csv("../Evaluation_Data/Minimax/mm_d4.csv")
mm_depth6 = pd.read_csv("../Evaluation_Data/Minimax/mm_d6.csv")
```

```python
print(ab_depth2.describe())
print(mm_depth2.describe())
```

```python
import matplotlib.pyplot as plt
```

# Minimax Vs Alpha-Beta

## average number of state evaluated per game

```python
alg    = ['Minimax-D2', 'AlphaBeta-D2',
          'Minimax-D4', 'AlphaBeta-D4',
          'Minimax-D6', 'AlphaBeta-D6',]
state_evaluated_values = [mm_depth2['state_evaluated'].mean(), ab_depth2['state_evaluated'].mean(), 
                          mm_depth4['state_evaluated'].mean(), ab_depth4['state_evaluated'].mean(), 
                          mm_depth6['state_evaluated'].mean(), ab_depth6['state_evaluated'].mean()]

bars = plt.bar(alg, state_evaluated_values)

plt.title('Rerata Jumlah State Dievaluasi Dalam 1 Game')
plt.xlabel('Algoritma')
plt.xticks(rotation=45)
plt.ylabel('State Dievaluasi')

plt.bar_label(bars, padding=3, fmt='%.2f')

plt.show()
```

## average number of state pruned per game

```python
alg    = ['Minimax-D2', 'AlphaBeta-D2',
          'Minimax-D4', 'AlphaBeta-D4',
          'Minimax-D6', 'AlphaBeta-D6',]
pruned_values = [mm_depth2['state_pruned'].mean(), ab_depth2['state_pruned'].mean(), 
                 mm_depth4['state_pruned'].mean(), ab_depth4['state_pruned'].mean(), 
                 mm_depth6['state_pruned'].mean(), ab_depth6['state_pruned'].mean()]

bars = plt.bar(alg, pruned_values)

plt.title('Rerata jumlah state yang di-prune dalam 1 game')
plt.xlabel('Algoritma')
plt.xticks(rotation=45)
plt.ylabel('State di-prune')

plt.bar_label(bars, padding=3, fmt='%.2f')

plt.show()
```

## average time execution per game

```python
alg    = ['Minimax-D2', 'AlphaBeta-D2',
          'Minimax-D4', 'AlphaBeta-D4',
          'Minimax-D6', 'AlphaBeta-D6',]
time_values   = [mm_depth2['time_elapsed'].mean(), ab_depth2['time_elapsed'].mean(), 
                 mm_depth4['time_elapsed'].mean(), ab_depth4['time_elapsed'].mean(), 
                 mm_depth6['time_elapsed'].mean(), ab_depth6['time_elapsed'].mean()]

bars = plt.bar(alg, time_values)

plt.title('Rerata waktu yang dibutuhkan')
plt.xlabel('Algoritma')
plt.xticks(rotation=45)
plt.ylabel('Waktu (ms)')

plt.bar_label(bars, padding=3, fmt='%.2f')

plt.show()
```
